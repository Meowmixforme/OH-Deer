First let's import the database and take a look at it's structure

```python
import pandas as pd

df = pd.read_csv("records-2025-12-28.csv")

df.head()
df.info()
df.describe(include='all')
```

There are 28,964 rows and from a visual inspection we can see that:

- around 5,500 rows have no Event Date
- around 250 rows have no coordinates
- some records span date ranges
- some are duplicates from multiple datasets

First let's filter the record to only include valid spatial and temporal records.

```python
df = pd.read_csv("records-2025-12-28.csv")

# Parse dates
df['Event Date'] = pd.to_datetime(df['Event Date'], errors='coerce')

# Keep only rows with usable coordinates AND a valid date
df_clean = df.dropna(subset=['Latitude (WGS84)', 'Longitude (WGS84)', 'Event Date'])

len(df_clean)
```

Now we have 23422 records we should standardise column names

```python
df_clean = df_clean.rename(columns={
    'Latitude (WGS84)': 'lat',
    'Longitude (WGS84)': 'lon',
    'Coordinate Uncertainty (m)': 'uncertainty',
    'Event Date': 'date',
    'Scientific name': 'scientific_name',
    'Common name': 'common_name'
})
```

Checking the count by year

```python
pd.set_option('display.max_rows', None)
df_clean['year'].value_counts().sort_index()
```

We can see Sparse, scattered early records from 1705 to around 1950.

A small rise in the late 1950s–1960s.

A huge jump in 1970 and then stable, dense data from 1980 onward.

There is a massive recorder boom in the 2000s–2020s, likely from citizen science contributions.

For the project it would be best to use data from 1970 to 2024, as the dataset was downloaded in 2025 and that year is incomplete. This gives us a long enough timescale with enough density and spatial coverage.




Checking coordinate uncertainty distribution


```python
df_clean['uncertainty'].describe()
```

Almost every record has an uncertainty value. We see a mixture of modern, high‑quality submissions that are extremely precise. Half the dataset is within around 70m accuracy. Three‑quarters of the records are within around 700m accuracy.

The mean and standard deviation are inflated by a small number of very coarse records.

To be safe we should model at 1km resolution.

Dropping extremely coarse records

```python
df_mod = df_clean[df_clean['uncertainty'] <= 5000]
```

Checking the new number of records with coarse omitted

```python
len(df_mod)
```

Plotting the cleaned points

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,8))
plt.scatter(df_mod['lon'], df_mod['lat'], s=1, alpha=0.3)
plt.title("Cleaned Deer Occurrence Points")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.show()
```

Filtering to selected years

```python
df_mod = df_mod[(df_mod['year'] >= 1970) & (df_mod['year'] <= 2024)]
```

Dropping duplicate years

```python
df_mod = df_mod.drop_duplicates(subset=['lat', 'lon', 'date'])
```

Creating a GeoDataFrame in OSGB (British National Grid)

```python
import geopandas as gpd
from shapely.geometry import Point

gdf = gpd.GeoDataFrame(
    df_mod,
    geometry=gpd.points_from_xy(df_mod['lon'], df_mod['lat']),
    crs="EPSG:4326"
)

# Reproject to OSGB (units = metres)
gdf = gdf.to_crs("EPSG:27700")

# Create 1 km grid indices
gdf['x_1km'] = (gdf.geometry.x // 1000).astype(int)
gdf['y_1km'] = (gdf.geometry.y // 1000).astype(int)
```

Aggregate occurrences into 1km cells

```python
df_grid = (
    gdf
    .groupby(['x_1km', 'y_1km', 'year'])
    .size()
    .reset_index(name='count')
)
```

A visualisation

```python
plt.figure(figsize=(6,8))
plt.scatter(df_grid['x_1km'], df_grid['y_1km'], s=1, alpha=0.3)
plt.title("1km Grid Cells with Deer Records")
plt.xlabel("OSGB X (km)")
plt.ylabel("OSGB Y (km)")
plt.show()
```
